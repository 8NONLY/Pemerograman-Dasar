```python
# Jawaban Latihan 6

pembangkit = [
    {
        "nama": "PLTU Suralaya",
        "kapasitas_MW": 1000,
        "status": "operasi"
    },
    {
        "nama": "PLTA Cirata",
        "kapasitas_MW": 500,
        "status": "operasi"
    },
    {
        "nama": "PLTS Likupang",
        "kapasitas_MW": 100,
        "status": "pemeliharaan"
    }
]

# Cetak nama unit kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

# Ubah status salah satu unit
pembangkit[2]["status"] = "operasi"

# Tambahkan unit baru
pembangkit.append({
    "nama": "PLTB Sidrap",
    "kapasitas_MW": 75,
    "status": "operasi"
})

# Cetak seluruh data dalam format tabel
print("\nDaftar Pembangkit")
print("-" * 50)
print(f"{'Nama':<20} {'Kapasitas(MW)':<15} {'Status'}")
print("-" * 50)

for unit in pembangkit:
    print(
        f"{unit['nama']:<20} "
        f"{unit['kapasitas_MW']:<15} "
        f"{unit['status']}"
    )
```
