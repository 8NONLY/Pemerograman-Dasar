```python
# Jawaban Latihan 4

# Data awal
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir memakai indeks
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data dengan len()
print("Jumlah total data:", len(tegangan_harian))
```

```python
# Jawaban Latihan 5

motor_listrik = {
    "nama": "Motor Induksi",
    "daya_kW": 15,
    "tegangan_V": 380,
    "rpm": 1450
}

# Cetak salah satu nilai
print(motor_listrik["rpm"])

# Ubah nilai rpm
motor_listrik["rpm"] = 1500

# Cetak ulang seluruh dictionary
print(motor_listrik)
```

```python
# Jawaban Latihan 6

# Data awal (List of Dictionary)
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80, "status": "pemeliharaan"},
]

# 1. Cetak nama unit kedua (indeks ke-1)
print("Nama unit kedua:", pembangkit[1]["nama"])

# 2. Ubah status salah satu unit (contoh: ubah status Unit-3 menjadi "operasi")
pembangkit[2]["status"] = "operasi"

# 3. Tambahkan satu unit baru ke list dengan append()
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 120, "status": "pemeliharaan"})

# 4. Cetak seluruh data
print("\nSeluruh data pembangkit:")
for unit in pembangkit:
    print(unit)
```
