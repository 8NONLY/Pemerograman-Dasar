# 1. Urutan Prioritas Operator dalam Python

Operator precedence adalah aturan yang menentukan operator mana yang dikerjakan lebih dahulu dalam suatu ekspresi. Dalam Python, tanda kurung memiliki prioritas utama, diikuti perpangkatan, perkalian atau pembagian, penjumlahan atau pengurangan, operator perbandingan, kemudian operator logika not, and, dan or.

pada expresi berikut:

```python
10 + 5 * 2
```

hasilnya adalah 20, bukan 30, karena operator perkalian * memiliki prioritas lebih tinggi daripada penjumlahan +. Python menghitung 5 * 2 = 10 terlebih dahulu, kemudian menghitung 10 + 10 = 20.

Jika ingin menghasilkan 30, penjumlahan harus ditempatkan di dalam tanda kurung:

```python
(10 + 5) * 2
```

Python akan menghitung 10 + 5 = 15, kemudian 15 * 2 = 30. Dengan demikian, tanda kurung dapat digunakan untuk mengubah urutan perhitungan dan memperjelas maksud program.

#2. Perbedaan Operator Relasional dan Logika

Operator relasional digunakan untuk membandingkan dua nilai dan biasanya menghasilkan True atau False. Contohnya adalah >, , >=, <=, ==, dan !=. Sementara itu, operator logika digunakan untuk menggabungkan atau membalik beberapa kondisi, yaitu and, or, dan not.

Contoh penerapannya pada sistem pembangkit adalah sebagai berikut:

```python
suhu = 95
tekanan = 120

if suhu > 90 and tekanan > 110:
    print("Alarm aktif: kondisi melebihi batas aman.")
else:
    print("Kondisi masih aman.")
```

Pada kode tersebut, > merupakan operator relasional yang membandingkan suhu dan tekanan dengan batas aman. Sementara itu, and merupakan operator logika yang menggabungkan kedua kondisi. Alarm hanya aktif apabila suhu lebih dari 90 dan tekanan lebih dari 110. Karena kedua kondisi bernilai True, program menampilkan pesan bahwa kondisi telah melebihi batas aman.

source: https://docs.python.org/3/builtins/stdtypes.html

#3. Kasus Kesalahan Akibat Operator Precedence

Salah satu kesalahan yang sering dibahas di forum pemrograman adalah penggunaan operator and dan or tanpa tanda kurung. Dalam Python, and memiliki prioritas lebih tinggi daripada or, sehingga ekspresi A or B and C dibaca sebagai A or (B and C), bukan (A or B) and C. Hal tersebut dapat menyebabkan program memberikan keputusan yang tidak sesuai dengan maksud programmer. Perbaikannya adalah menggunakan tanda kurung untuk memperjelas kelompok kondisi yang harus diperiksa

Contoh kode yang salah:

```python
nama = "Hadi"
aktif = False

if nama == "Hadi" or nama == "Wibowo" and aktif:
    print("Akses diberikan.")
else:
    print("Akses ditolak.")
```

Python membaca kondisi tersebut sebagai Hadi aktif

Akibatnya, pengguna bernama Hadi tetap mendapatkan akses meskipun statusnya tidak aktif. Jika aturan yang dimaksud adalah Hadi atau Wibowo hanya boleh memperoleh akses ketika statusnya aktif, kode harus diperbaiki menjadi:

```python
if (nama == "Hadi" or nama == "Wibowo") and aktif:
    print("Akses diberikan.")
else:
    print("Akses ditolak.")
```

Pada kode yang telah diperbaiki, status aktif berlaku untuk kedua nama sehingga akses akan ditolak ketika pengguna tidak aktif.
