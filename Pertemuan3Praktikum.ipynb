#**B.1 Soal Umum**

```python
# Nama    : Muhamad Hadi Wibowo Mulyadi
# NIM     : 4232501060
# Tanggal : 24 September 2026
# Tujuan  : Menghitung hasil operasi aritmatika dari dua bilangan bulat

bilangan_1 = int(input("Masukkan bilangan bulat pertama: "))
bilangan_2 = int(input("Masukkan bilangan bulat kedua selain nol: "))

print(f"Hasil penjumlahan       : {bilangan_1 + bilangan_2}")
print(f"Hasil pengurangan       : {bilangan_1 - bilangan_2}")
print(f"Hasil perkalian         : {bilangan_1 * bilangan_2}")
print(f"Hasil pembagian         : {bilangan_1 / bilangan_2}")
print(f"Hasil pembagian bulat   : {bilangan_1 // bilangan_2}")
print(f"Hasil modulus atau sisa : {bilangan_1 % bilangan_2}")
```

```python
# Nama    : Muhamad Hadi Wibowo Mulyadi
# NIM     : 4232501060
# Tanggal : 24 September 2026
# Tujuan  : Menentukan apakah nilai ujian termasuk kategori lulus

nilai_ujian = float(input("Masukkan nilai ujian (0-100): "))

lulus = nilai_ujian >= 60

print(f"Nilai ini termasuk kategori lulus (>= 60): {lulus}")
```

```python
# Nama    : Muhamad Hadi Wibowo Mulyadi
# NIM     : 4232501060
# Tanggal : 24 September 2026
# Tujuan  : Menentukan apakah seseorang boleh mengemudi sendiri

umur = int(input("Masukkan umur: "))
input_sim = input("Apakah memiliki SIM? (True/False): ")

punya_sim = input_sim.strip().lower() == "true"

boleh_mengemudi_sendiri = umur >= 17 and punya_sim

# Tanda kurung diperlukan agar operator not membalik seluruh kondisi
tidak_boleh_mengemudi_sendiri = not (umur >= 17 and punya_sim)

print(f"Boleh mengemudi sendiri: {boleh_mengemudi_sendiri}")
print(f"Tidak boleh mengemudi sendiri: {tidak_boleh_mengemudi_sendiri}")
```

#**B.2 Soal Konteks Energi**

```python
# Nama    : Muhamad Hadi Wibowo Mulyadi
# NIM     : 4232501060
# Tanggal : 24 September 2026
# Tujuan  : Menghitung efisiensi termal sebuah unit pembangkit

energi_input = float(input("Masukkan energi input (MJ): "))
energi_output = float(input("Masukkan energi output (MJ): "))

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")
```

```python
# Nama    : Muhamad Hadi Wibowo Mulyadi
# NIM     : 4232501060
# Tanggal : 24 September 2026
# Tujuan  : Menentukan apakah unit pembangkit perlu maintenance

suhu_boiler = float(input("Masukkan suhu boiler (°C): "))
beban_operasi = float(input("Masukkan beban operasi (%): "))

perlu_maintenance = suhu_boiler > 550 or beban_operasi > 95

print(
    f"Pada suhu boiler {suhu_boiler:.2f}°C dan beban operasi "
    f"{beban_operasi:.2f}%, status perlu maintenance adalah "
    f"{perlu_maintenance}."
)
```
